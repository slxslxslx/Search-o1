```python
import json
import os

def clean_json(input_path, output_path=None):
    """
    读取JSON文件，删除每条记录的Metrics字段，只保留id, Question, answer, Output。
    
    Args:
        input_path (str): 输入JSON文件路径
        output_path (str): 输出JSON文件路径，若为None则覆盖原文件（请谨慎）
    """
    # 读取原始数据
    with open(input_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # 检查是否为列表
    if not isinstance(data, list):
        raise ValueError("JSON文件根元素必须是数组（列表）")
    
    # 清理每条记录
    cleaned_data = []
    for item in data:
        # 只保留指定字段，若字段不存在则使用默认值（此处设为None，但通常都存在）
        cleaned_item = {
            "id": item.get("id"),
            "Question": item.get("Question"),
            "answer": item.get("answer"),
            "Output": item.get("Output")
        }
        cleaned_data.append(cleaned_item)
    
    # 确定输出路径
    if output_path is None:
        output_path = input_path  # 覆盖原文件（请确保备份）
    
    # 写入新文件
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(cleaned_data, f, indent=2, ensure_ascii=False)
    
    print(f"清理完成！共处理 {len(cleaned_data)} 条记录。")
    print(f"输出文件：{output_path}")

if __name__ == "__main__":
    # # 请根据实际需要修改路径
    # input_file = "/home/shaolingxuan/project/Search-o1/outputs/runs.baselines/bamboogle.qwen3.5-9b.search_o1/all.json"
    # # 建议输出到新文件，避免误覆盖
    # output_file = "/home/shaolingxuan/project/Search-o1/outputs/runs.baselines/bamboogle.qwen3.5-9b.search_o1/all_cleaned.json"
    
    # 请根据实际需要修改路径
    input_file = "/home/shaolingxuan/project/Search-o1/outputs/runs.baselines/seal0.qwen3.5-9b.search_o1/all.json"
    # 建议输出到新文件，避免误覆盖
    output_file = "/home/shaolingxuan/project/Search-o1/outputs/runs.baselines/seal0.qwen3.5-9b.search_o1/all_cleaned.json"
    
    clean_json(input_file, output_file)
```
